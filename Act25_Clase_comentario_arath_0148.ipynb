En este ejercicio, la clase Comentario no solo sirve para almacenar información, sino que también gestiona su propia entrada y salida de datos a través de métodos.

leer_datos(): Es una función de interacción que permite capturar la información del usuario en tiempo real y asignarla a las propiedades del objeto.

mostrar_datos(): Es una función de salida que organiza y presenta la información almacenada de forma legible.

Al utilizar dos objetos (coment1 y coment2), demostramos la capacidad de la Programación Orientada a Objetos para manejar múltiples registros de manera estructurada, asegurando que el comentario de un usuario no sobrescriba ni interfiera con el del otro.

![texto del vínculo](https://raw.githubusercontent.com/JacoboValenzuela128/Act25-Clase-comentario-Arath-0148/refs/heads/main/infoclasecomentario.png)

```python
class Comentario:
    # El constructor inicializa los atributos (pueden empezar vacíos)
    def __init__(self, usuario="", texto="", fecha=""):
        self.usuario = usuario
        self.texto = texto
        self.fecha = fecha

    # Función para pedir los datos al usuario
    def leer_datos(self):
        print(f"\n--- Escribiendo nuevo comentario ---")
        self.usuario = input("Nombre de usuario: ")
        self.texto = input("Escribe tu comentario: ")
        self.fecha = input("Fecha (ej. 28/04/2024): ")

    # Función para imprimir los datos en pantalla
    def mostrar_datos(self):
        print(f"\n💬 Usuario: {self.usuario}")
        print(f"📅 Fecha: {self.fecha}")
        print(f"📝 Comentario: {self.texto}")
        print("-" * 25)

# --- Creación de los 2 objetos ---

# Creamos los objetos (instancias)
coment1 = Comentario()
coment2 = Comentario()

# --- Ejecución de las funciones ---

# El objeto 1 pide y muestra sus datos
coment1.leer_datos()
coment1.mostrar_datos()

# El objeto 2 pide y muestra sus datos
coment2.leer_datos()
coment2.mostrar_datos()

print("\n-------------------------------------------------------------------")
print(" Programa interactivo creado por Arath Valenzuela (NL52) & Gemini AI")
```
